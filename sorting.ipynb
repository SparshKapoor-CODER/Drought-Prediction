```python
import pandas as pd
import os
import matplotlib.pyplot as plt

print("libraries imported")

# Load the data
data = pd.read_csv('forest_loss_30m_risk - forest_loss_30m_risk.csv')
print("data loaded")

print(data.head(10))  # Display the first few rows of the dataset

os.makedirs('state_data', exist_ok=True)  # Create a directory to store state data files
```

```python
# create Csv file of each state with the data of that state in the directory state_data
for state in data['state'].unique():
    state_data = data[data['state'] == state]
    state_data.to_csv(f'state_data/{state}_data.csv', index=False)
print("state data files created")
```
